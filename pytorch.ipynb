```python
# عندنا وزن اسمه \(w\)، وعايزين نعرف لو غيّرنا قيمته، الـ loss هيتغير إزاي.
w = torch.tensor([2.0],requires_grad=True)

x = torch.tensor([3.0])

loss = (w * x-1)** 2

loss.backward()
```

```python
# Neural Network Model بداية إنشاء  
import torch.nn as nn # بتتنفذ كا كلاس 


# وظائفه الاساسيه  PyTorch  و ده بيتيح ليه nn.Module بيورث من موديل اسمه  TinyModel معنها ان 
class TinyModel(nn.Module):
    def __init__(self):
        super().__init__()

        # self.w = torch.tensor([2.0],requires_grad=True) # Gradient يحسب لها 
        # self.b = torch.tensor([0.0],requires_grad=True) # # Gradient يحسب لها 

        #requires_grad فيه طريقه تاني للكتابه تغنينا عن 

        self.w =nn.Parameter(torch.tensor([2.0]))
        self.b =nn.Parameter(torch.tensor([0.0]))

    def forward(self,x):
      return (self.w *x ) + self.b    # Prediction
```

```python
# All Of Code 

# Neural Network Model بداية إنشاء  
import torch.nn as nn # بتتنفذ كا كلاس 


# وظائفه الاساسيه  PyTorch  و ده بيتيح ليه nn.Module بيورث من موديل اسمه  TinyModel معنها ان 
class TinyModel(nn.Module):
    def __init__(self):
        super().__init__()

        # self.w = torch.tensor([2.0],requires_grad=True) # Gradient يحسب لها 
        # self.b = torch.tensor([0.0],requires_grad=True) # # Gradient يحسب لها 

        #requires_grad فيه طريقه تاني للكتابه تغنينا عن 

        self.w =nn.Parameter(torch.tensor([2.0]))
        self.b =nn.Parameter(torch.tensor([0.0]))

    def forward(self,x):
      return (self.w *x ) + self.b    # Prediction


model =TinyModel()
x=torch.tensor([3.0])    
target=torch.tensor([1.0])    
eta = 0.01

for step in range(5):
   # 1 Forword
   prediction=model(x)

   # 2 Loss
   loss = (prediction-target) ** 2 

   # 3 Backword
   loss.backward()


   print(
        f"Step={step}",
        f"Prediction={prediction.item():.2f}", # item() عشان نحول تينسور الي رقم عادي 
        f"Loss={loss.item():.2f} ",
        f"w.grad={model.w.grad.item():.2f}"
   )






   # 4 Update
   with torch.no_grad():
      model.w -=eta * model.w.grad
      model.b -=eta * model.b.grad

    # 5 Clear Old Gradient
   model.w.grad.zero_()    
   model.b.grad.zero_()    #هيتذبذب  loss لو خذفت ده 
```
